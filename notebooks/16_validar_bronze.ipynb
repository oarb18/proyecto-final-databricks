```python

# ============================================================
# ELECTROCASA - CONTROL DE FINALIZACION BRONZE
# ============================================================

tablas = [
    "ventas",
    "catalogo",
    "sucursales",
    "empleados",
    "resenas",
    "devoluciones",
    "tracking_envios"
]

for nombre in tablas:
    tabla = f"proyecto_final.bronze.{nombre}"

    if not spark.catalog.tableExists(tabla):
        raise Exception(f"No existe la tabla: {tabla}")

    total = spark.table(tabla).count()

    if total == 0:
        raise Exception(f"Tabla Bronze vacia: {tabla}")

    print(f"OK - {tabla}: {total} registros")

print("CONTROL BRONZE FINALIZADO CORRECTAMENTE")
```
