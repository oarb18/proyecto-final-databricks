```python

# ============================================================
# PROYECTO FINAL ELECTROCASA
# 17. VALIDACION DE FINALIZACION SILVER
# ============================================================

CATALOGO = "proyecto_final"

FUENTES = [
    "ventas",
    "catalogo",
    "sucursales",
    "empleados",
    "resenas",
    "devoluciones",
    "tracking_envios"
]

errores = []
resultados = []

print("VALIDACION DE LAS 7 FUENTES SILVER")
print("=" * 60)

for fuente in FUENTES:
    tabla = f"{CATALOGO}.silver.{fuente}"

    if not spark.catalog.tableExists(tabla):
        errores.append(f"No existe: {tabla}")
        print(f"ERROR | {tabla} | No existe")
        continue

    total = spark.table(tabla).count()
    resultados.append((fuente, total))

    if total == 0:
        errores.append(f"Tabla vacia: {tabla}")
        print(f"ERROR | {tabla} | 0 registros")
    else:
        print(f"OK | {tabla} | {total} registros")

print("=" * 60)

if errores:
    raise RuntimeError(
        "VALIDACION SILVER FALLIDA:\n" + "\n".join(errores)
    )

print("SILVER VALIDADO: 7 FUENTES DISPONIBLES")
```
