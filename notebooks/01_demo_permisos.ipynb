```python
# Variables de la demostración
CATALOGO = "proyecto_final"
GRUPO_ANA = "`electrocasa_analistas`"
```

```python
# ============================================================
# 08. DEMOSTRACIÓN DE GRANT Y REVOKE
# Proyecto: ElectroCasa
# ============================================================

# Otorgar temporalmente un permiso indebido
spark.sql(f"""
    GRANT SELECT
    ON SCHEMA {CATALOGO}.silver
    TO {GRUPO_ANA}
""")

print("ANTES DEL REVOKE:")

display(
    spark.sql(f"""
        SHOW GRANTS ON SCHEMA {CATALOGO}.silver
    """)
)
```

```python
%sql
REVOKE SELECT
ON SCHEMA proyecto_final.silver
FROM `electrocasa_analistas`;
```

```python
%sql
SHOW GRANTS ON SCHEMA proyecto_final.silver;
```
